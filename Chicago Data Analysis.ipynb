## Load CSV Data into SQLite Tables

```python
# Load Census Data
census_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DB0201EN-SkillsNetwork/labs/FinalModule_Coursera_V5/data/ChicagoCensusData.csv'
df_census = pd.read_csv(census_url)
df_census.to_sql('CENSUS_DATA', conn, if_exists='replace', index=False)

print(f"✓ CENSUS_DATA table created with {len(df_census)} rows")
df_census.head(3)
```

```python
# Load Public Schools Data
schools_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DB0201EN-SkillsNetwork/labs/FinalModule_Coursera_V5/data/ChicagoPublicSchools.csv'
df_schools = pd.read_csv(schools_url)
df_schools.to_sql('CHICAGO_PUBLIC_SCHOOLS', conn, if_exists='replace', index=False)

print(f"✓ CHICAGO_PUBLIC_SCHOOLS table created with {len(df_schools)} rows")
df_schools.head(3)
```

# SQL Analysis - 10 Problems

Now let's solve the 10 assignment problems using SQL queries.

---
## Problem 1: Total Number of Crimes
**Question:** Find the total number of crimes recorded in the CRIME table.

```python
%%sql
SELECT COUNT(*) AS TOTAL_CRIMES
FROM CHICAGO_CRIME_DATA;
```

---
## Problem 2: Community Areas with Low Income
**Question:** List community area names and numbers with per capita income less than 11000.

```python
%%sql
SELECT COMMUNITY_AREA_NUMBER, COMMUNITY_AREA_NAME
FROM CENSUS_DATA
WHERE PER_CAPITA_INCOME_ < 11000;
```

---
## Problem 3: Crimes Involving Minors
**Question:** List all case numbers for crimes involving minors. (Note: Children are not considered minors for crime analysis purposes)

```python
%%sql
SELECT CASE_NUMBER, PRIMARY_TYPE, DESCRIPTION
FROM CHICAGO_CRIME_DATA
WHERE DESCRIPTION LIKE '%MINOR%';
```

---
## Problem 4: Kidnapping Crimes Involving Children
**Question:** List all kidnapping crimes involving a child.

```python
%%sql
SELECT CASE_NUMBER, PRIMARY_TYPE, DESCRIPTION, DATE, LOCATION_DESCRIPTION
FROM CHICAGO_CRIME_DATA
WHERE PRIMARY_TYPE = 'KIDNAPPING'
  AND (DESCRIPTION LIKE '%CHILD%' OR DESCRIPTION LIKE '%MINOR%');
```

---
## Problem 5: Types of Crimes at Schools
**Question:** List the types of crimes recorded at schools (no repetitions).

```python
%%sql
SELECT DISTINCT PRIMARY_TYPE
FROM CHICAGO_CRIME_DATA
WHERE LOCATION_DESCRIPTION LIKE '%SCHOOL%'
ORDER BY PRIMARY_TYPE;
```

---
## Problem 6: Average Safety Score by School Type
**Question:** List the type of schools along with the average safety score for each type.

```python
%%sql
SELECT "School_Type" AS SCHOOL_TYPE,
       AVG(CAST("Safety_Score" AS REAL)) AS AVERAGE_SAFETY_SCORE
FROM CHICAGO_PUBLIC_SCHOOLS
WHERE "Safety_Score" IS NOT NULL AND "Safety_Score" != 'NDA'
GROUP BY "School_Type";
```

---
## Problem 7: Top 5 Community Areas with Highest Poverty
**Question:** List 5 community areas with highest percentage of households below poverty line.

```python
%%sql
SELECT COMMUNITY_AREA_NUMBER, COMMUNITY_AREA_NAME, PERCENT_HOUSEHOLDS_BELOW_POVERTY
FROM CENSUS_DATA
ORDER BY PERCENT_HOUSEHOLDS_BELOW_POVERTY DESC
LIMIT 5;
```

---
## Problem 8: Most Crime Prone Community Area
**Question:** Which community area is most crime prone? Display the community area number only.

```python
%%sql
SELECT COMMUNITY_AREA_NUMBER, COUNT(*) AS CRIME_COUNT
FROM CHICAGO_CRIME_DATA
WHERE COMMUNITY_AREA_NUMBER IS NOT NULL
GROUP BY COMMUNITY_AREA_NUMBER
ORDER BY CRIME_COUNT DESC
LIMIT 1;
```

---
## Problem 9: Community Area with Highest Hardship Index (Using Subquery)
**Question:** Use a sub-query to find the name of the community area with highest hardship index.

```python
%%sql
SELECT COMMUNITY_AREA_NAME
FROM CENSUS_DATA
WHERE HARDSHIP_INDEX = (
    SELECT MAX(HARDSHIP_INDEX)
    FROM CENSUS_DATA
);
```

---
## Problem 10: Community Area with Most Crimes (Using Subquery)
**Question:** Use a sub-query to determine the Community Area Name with most number of crimes.

```python
%%sql
SELECT COMMUNITY_AREA_NAME
FROM CENSUS_DATA
WHERE COMMUNITY_AREA_NUMBER = (
    SELECT COMMUNITY_AREA_NUMBER
    FROM CHICAGO_CRIME_DATA
    WHERE COMMUNITY_AREA_NUMBER IS NOT NULL
    GROUP BY COMMUNITY_AREA_NUMBER
    ORDER BY COUNT(*) DESC
    LIMIT 1
);
```

# Bonus: Data Visualizations

Let's create some visualizations to better understand the data.

## Visualization 1: Top 10 Community Areas by Crime Count

```python
# Query top 10 areas by crime
query = """
SELECT cd.COMMUNITY_AREA_NAME, COUNT(ccd.ID) AS CRIME_COUNT
FROM CHICAGO_CRIME_DATA ccd
JOIN CENSUS_DATA cd ON ccd.COMMUNITY_AREA_NUMBER = cd.COMMUNITY_AREA_NUMBER
GROUP BY cd.COMMUNITY_AREA_NAME
ORDER BY CRIME_COUNT DESC
LIMIT 10
"""
df_top_crime = pd.read_sql_query(query, conn)

# Create plot
plt.figure(figsize=(12, 6))
plt.barh(df_top_crime['COMMUNITY_AREA_NAME'][::-1], df_top_crime['CRIME_COUNT'][::-1], color='crimson')
plt.xlabel('Number of Crimes', fontsize=12)
plt.ylabel('Community Area', fontsize=12)
plt.title('Top 10 Community Areas by Crime Count', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Visualization 2: Hardship Index vs Per Capita Income

```python
# Load census data for visualization
df_census_viz = pd.read_sql_query("SELECT * FROM CENSUS_DATA", conn)

# Create scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(df_census_viz['PER_CAPITA_INCOME_'], df_census_viz['HARDSHIP_INDEX'], 
            alpha=0.6, s=100, c=df_census_viz['HARDSHIP_INDEX'], cmap='YlOrRd')
plt.xlabel('Per Capita Income ($)', fontsize=12)
plt.ylabel('Hardship Index', fontsize=12)
plt.title('Relationship: Per Capita Income vs Hardship Index', fontsize=14, fontweight='bold')
plt.colorbar(label='Hardship Index')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## Visualization 3: Crime Types Distribution

```python
# Query top crime types
query = """
SELECT PRIMARY_TYPE, COUNT(*) AS COUNT
FROM CHICAGO_CRIME_DATA
GROUP BY PRIMARY_TYPE
ORDER BY COUNT DESC
LIMIT 10
"""
df_crime_types = pd.read_sql_query(query, conn)

# Create pie chart
plt.figure(figsize=(10, 8))
plt.pie(df_crime_types['COUNT'], labels=df_crime_types['PRIMARY_TYPE'], 
        autopct='%1.1f%%', startangle=90, colors=plt.cm.Set3.colors)
plt.title('Top 10 Crime Types Distribution', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
